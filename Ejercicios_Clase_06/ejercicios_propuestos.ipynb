```scala
// Definición de variables
val nombre: String = "Mariano Navarro"
var edad: Int = 50
val altura: Double = 1.88
var estaTrabajando: Boolean = false

// Impresión con interpolador s
println(s"""=== Ficha Personal ===
Nombre:   $nombre
Edad:     $edad años
Altura:   $altura m
Trabajando: $estaTrabajando""")
```

## val (Valores inmutables):

### nombre:
    Generalmente, el nombre de una persona en un contexto de ficha técnica no cambia durante la ejecución del programa. Usar val hace que el código sea más seguro y predecible.

### altura:
    Aunque una persona puede crecer (o encogerse un poco con la edad), en el contexto de una ficha estática se considera un valor fijo. Si quisiéramos seguir su evolución médica, usaríamos var, pero para una ficha de perfil, val es más robusto.

## var (Variables mutables):

## edad:
    Es un valor que cambia inevitablemente con el paso del tiempo. Si esta aplicación fuera parte de un sistema que persiste en el tiempo, necesitaríamos modificar este campo cada año.

## estaTrabajando:
    El estado laboral es dinámico. Una persona puede ser contratada o dejar su puesto en cualquier momento, por lo que declarar esta variable como var permite que el programa refleje cambios en la situación real del usuario sin necesidad de crear un objeto nuevo.
