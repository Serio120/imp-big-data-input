### 1. Ejemplo propio de __case class__

```scala
case class Libro(titulo: String, autor: String, paginas: Int)

val quijote = Libro("Don Quijote de la Mancha", "Miguel de Cervantes", 863)
val copiaQuijote = Libro("Don Quijote de la Mancha", "Miguel de Cervantes", 863)

println(quijote)
println(quijote == copiaQuijote)

val quijoteCorto = quijote.copy(paginas = 300)
println(quijoteCorto)
println(quijote)
```

### Explicación Línea a Línea

- **`case class Libro(titulo: String, autor: String, paginas: Int)`**  
  Defines una `case class` llamada `Libro` con tres campos inmutables: `titulo`, `autor` y `paginas`. Scala genera automáticamente `toString`, `equals`, `hashCode` y `copy`, además del constructor sin `new`.

- **`val quijote = Libro("Don Quijote de la Mancha", "Miguel de Cervantes", 863)`**  
  Crea una instancia de `Libro` usando el constructor generado, sin `new`. Representa un libro concreto.

- **`val copiaQuijote = Libro("Don Quijote de la Mancha", "Miguel de Cervantes", 863)`**  
  Crea otra instancia con los mismos datos, para demostrar la comparación por valor.

- **`println(quijote)`**  
  Imprime el libro usando el `toString` automático, algo como `Libro(Don Quijote de la Mancha,Miguel de Cervantes,863)`.

- **`println(quijote == copiaQuijote)`**  
  Compara ambos libros por contenido, no por referencia. Devuelve `true` porque los campos son iguales.

- **`val quijoteCorto = quijote.copy(paginas = 300)`**  
  Usa `copy` para crear un nuevo `Libro` con el mismo título y autor, pero con `paginas` distinto. El original no se modifica.

- **`println(quijoteCorto)`**  
  Muestra la versión “recortada” del libro, con 300 páginas.

- **`println(quijote)`**  
  Confirma que el objeto original sigue intacto, con 863 páginas.


### 2. Ejemplo propio de __case object__

```scala
case object Conectado
case object Desconectado
case object EnMantenimiento

val estadoServidor = Conectado

println(estadoServidor)

def mensajeEstado(estado: Any): String = estado match {
  case Conectado       => "El servidor está disponible."
  case Desconectado    => "El servidor está apagado."
  case EnMantenimiento => "El servidor está en mantenimiento."
  case _               => "Estado desconocido."
}

println(mensajeEstado(estadoServidor))
println(mensajeEstado(EnMantenimiento))
```

### Explicación Línea a Línea

- **`case object Conectado`**  
  Declara un `case object` singleton llamado `Conectado`. No tiene datos, solo representa un estado concreto.

- **`case object Desconectado`**  
  Otro estado posible del servidor, también como singleton.

- **`case object EnMantenimiento`**  
  Tercer estado posible, sin datos asociados.

- **`val estadoServidor = Conectado`**  
  Asigna el estado actual del servidor al valor `Conectado`.

- **`println(estadoServidor)`**  
  Imprime `Conectado` gracias al `toString` automático de `case object`.

- **`def mensajeEstado(estado: Any): String = estado match {`**  
  Define una función que recibe un valor de tipo `Any` y devuelve un `String`. Usa pattern matching para decidir el mensaje.

- **`case Conectado       => "El servidor está disponible."`**  
  Si el valor es exactamente el singleton `Conectado`, devuelve este mensaje.

- **`case Desconectado    => "El servidor está apagado."`**  
  Mensaje para el estado `Desconectado`.

- **`case EnMantenimiento => "El servidor está en mantenimiento."`**  
  Mensaje específico para `EnMantenimiento`.

- **`case _               => "Estado desconocido."`**  
  Caso comodín para cualquier otro valor que no coincida con los anteriores.

- **`println(mensajeEstado(estadoServidor))`**  
  Llama a la función con `Conectado`, imprime el mensaje correspondiente.

- **`println(mensajeEstado(EnMantenimiento))`**  
  Llama a la función con `EnMantenimiento`, imprime el mensaje de mantenimiento.


### 3. Ejemplo propio de pattern matching avanzado

#### 3.1 Desestructuración de `case class` + guards

```scala
case class Temperatura(ciudad: String, grados: Double)

def clasificarTemperatura(t: Temperatura): String = t match {
  case Temperatura(_, g) if g <= 0   => "Hace mucho frío."
  case Temperatura(_, g) if g <= 15  => "Hace fresco."
  case Temperatura(_, g) if g <= 25  => "Temperatura agradable."
  case Temperatura(_, g) if g <= 35  => "Hace calor."
  case Temperatura(_, _)             => "Hace muchísimo calor."
}

val t1 = Temperatura("Madrid", 8.0)
val t2 = Temperatura("Sevilla", 32.0)
val t3 = Temperatura("Reikiavik", -5.0)

println(clasificarTemperatura(t1))
println(clasificarTemperatura(t2))
println(clasificarTemperatura(t3))
```

### Explicación Línea a Línea

- **`case class Temperatura(ciudad: String, grados: Double)`**  
  Define una `case class` con el nombre de la ciudad y los grados.

- **`def clasificarTemperatura(t: Temperatura): String = t match {`**  
  Función que recibe una `Temperatura` y devuelve una descripción textual.

- **`case Temperatura(_, g) if g <= 0   => "Hace mucho frío."`**  
  Desestructura el objeto: ignora la ciudad (`_`) y captura los grados en `g`. Con el guard `if g <= 0` clasifica temperaturas bajo cero.

- **`case Temperatura(_, g) if g <= 15  => "Hace fresco."`**  
  Mismo patrón, pero para temperaturas entre 0 y 15.

- **`case Temperatura(_, g) if g <= 25  => "Temperatura agradable."`**  
  Clasifica temperaturas templadas.

- **`case Temperatura(_, g) if g <= 35  => "Hace calor."`**  
  Clasifica temperaturas cálidas.

- **`case Temperatura(_, _)             => "Hace muchísimo calor."`**  
  Cubre cualquier otra temperatura superior a 35 grados.

- **`val t1 = Temperatura("Madrid", 8.0)`**, **`t2`**, **`t3`**  
  Crea tres ejemplos de temperaturas en distintas ciudades.

- **`println(clasificarTemperatura(t1))`**, etc.  
  Imprime la clasificación de cada temperatura según las reglas anteriores.

#### 3.2 Matching sobre tipos

```scala
case class MensajeTexto(contenido: String)
case class MensajeImagen(url: String)
case class MensajeVideo(url: String, duracionSegundos: Int)

def procesarMensaje(m: Any): String = m match {
  case MensajeTexto(c)          => s"Texto recibido: $c"
  case MensajeImagen(u)         => s"Imagen recibida desde: $u"
  case MensajeVideo(u, d) if d <= 30 =>
    s"Video corto recibido desde: $u"
  case MensajeVideo(u, d)       =>
    s"Video largo de $d segundos desde: $u"
  case _                        => "Tipo de mensaje desconocido."
}

println(procesarMensaje(MensajeTexto("Hola")))
println(procesarMensaje(MensajeImagen("http://imagen.com/foto.png")))
println(procesarMensaje(MensajeVideo("http://video.com/clip.mp4", 20)))
println(procesarMensaje(MensajeVideo("http://video.com/pelicula.mp4", 1200)))
```

### Explicación Línea a Línea


- **`case class MensajeTexto(contenido: String)`**  
  Representa un mensaje de texto con su contenido.

- **`case class MensajeImagen(url: String)`**  
  Representa un mensaje que contiene una imagen, identificada por una URL.

- **`case class MensajeVideo(url: String, duracionSegundos: Int)`**  
  Representa un mensaje de video con URL y duración.

- **`def procesarMensaje(m: Any): String = m match {`**  
  Función que recibe “cualquier cosa” y decide qué hacer según el tipo real.

- **`case MensajeTexto(c)          => s"Texto recibido: $c"`**  
  Si `m` es un `MensajeTexto`, extrae el contenido en `c` y construye una respuesta.

- **`case MensajeImagen(u)         => s"Imagen recibida desde: $u"`**  
  Si es `MensajeImagen`, extrae la URL y la incluye en el mensaje.

- **`case MensajeVideo(u, d) if d <= 30 => ...`**  
  Si es un `MensajeVideo` y la duración es de 30 segundos o menos, lo considera un video corto.

- **`case MensajeVideo(u, d)       => ...`**  
  Cualquier otro `MensajeVideo` se trata como video largo.

- **`case _                        => "Tipo de mensaje desconocido."`**  
  Cubre cualquier otro tipo no contemplado.

- **Las llamadas a `println(procesarMensaje(...))`**  
  Demuestran cómo se comporta la función con cada tipo de mensaje.


### 4. Ejemplo propio de `sealed trait` + `case class` / `case object` (ADT)

```scala
sealed trait Notificacion

case class Email(destinatario: String, asunto: String, cuerpo: String) extends Notificacion
case class SMS(numero: String, texto: String)                          extends Notificacion
case object PushSilenciosa                                             extends Notificacion

def enviar(notificacion: Notificacion): Unit = notificacion match {
  case Email(dest, asu, _) =>
    println(s"Enviando email a $dest con asunto: $asu")
  case SMS(num, txt) =>
    println(s"Enviando SMS al número $num con texto: $txt")
  case PushSilenciosa =>
    println("Enviando notificación push silenciosa.")
}

val n1: Notificacion = Email("user@example.com", "Bienvenido", "Gracias por registrarte.")
val n2: Notificacion = SMS("+34123456789", "Tu código es 1234")
val n3: Notificacion = PushSilenciosa

enviar(n1)
enviar(n2)
enviar(n3)
```

### Explicación Línea a Línea

- **`sealed trait Notificacion`**  
  Declara un trait sellado. Todas las variantes de `Notificacion` deben estar en el mismo fichero. El compilador puede saber exactamente qué subtipos existen.

- **`case class Email(... ) extends Notificacion`**  
  Variante de notificación que lleva datos: destinatario, asunto y cuerpo. Es una `case class` que extiende `Notificacion`.

- **`case class SMS(numero: String, texto: String) extends Notificacion`**  
  Otra variante con datos: número de teléfono y texto del mensaje.

- **`case object PushSilenciosa extends Notificacion`**  
  Variante sin datos, representada como `case object`. Es una notificación push sin contenido visible.

- **`def enviar(notificacion: Notificacion): Unit = notificacion match {`**  
  Función que recibe una `Notificacion` y usa pattern matching para decidir cómo enviarla.

- **`case Email(dest, asu, _) =>`**  
  Si la notificación es un `Email`, extrae destinatario y asunto (ignora el cuerpo con `_`) y muestra un mensaje de envío.

- **`case SMS(num, txt) =>`**  
  Si es un `SMS`, extrae número y texto y muestra el envío correspondiente.

- **`case PushSilenciosa =>`**  
  Si es la variante `PushSilenciosa`, imprime un mensaje genérico.

- **`val n1: Notificacion = Email(...)`**, **`n2`**, **`n3`**  
  Crea tres notificaciones distintas, todas tipadas como `Notificacion` gracias al trait sellado.

- **`enviar(n1)`**, **`enviar(n2)`**, **`enviar(n3)`**  
  Llama a la función `enviar` con cada variante y ejecuta el caso correspondiente del `match`.
