### 1. Ejemplo propio de __case class__

```scala
case class Libro(titulo: String, autor: String, paginas: Int)

val quijote = Libro("Don Quijote de la Mancha", "Miguel de Cervantes", 863)
val copiaQuijote = Libro("Don Quijote de la Mancha", "Miguel de Cervantes", 863)

println(quijote)
println(quijote == copiaQuijote)

val quijoteCorto = quijote.copy(paginas = 300)
println(quijoteCorto)
println(quijote)
```

### Explicación Línea a Línea

- **`case class Libro(titulo: String, autor: String, paginas: Int)`**  
  Defines una `case class` llamada `Libro` con tres campos inmutables: `titulo`, `autor` y `paginas`. Scala genera automáticamente `toString`, `equals`, `hashCode` y `copy`, además del constructor sin `new`.

- **`val quijote = Libro("Don Quijote de la Mancha", "Miguel de Cervantes", 863)`**  
  Crea una instancia de `Libro` usando el constructor generado, sin `new`. Representa un libro concreto.

- **`val copiaQuijote = Libro("Don Quijote de la Mancha", "Miguel de Cervantes", 863)`**  
  Crea otra instancia con los mismos datos, para demostrar la comparación por valor.

- **`println(quijote)`**  
  Imprime el libro usando el `toString` automático, algo como `Libro(Don Quijote de la Mancha,Miguel de Cervantes,863)`.

- **`println(quijote == copiaQuijote)`**  
  Compara ambos libros por contenido, no por referencia. Devuelve `true` porque los campos son iguales.

- **`val quijoteCorto = quijote.copy(paginas = 300)`**  
  Usa `copy` para crear un nuevo `Libro` con el mismo título y autor, pero con `paginas` distinto. El original no se modifica.

- **`println(quijoteCorto)`**  
  Muestra la versión “recortada” del libro, con 300 páginas.

- **`println(quijote)`**  
  Confirma que el objeto original sigue intacto, con 863 páginas.


### 2. Ejemplo propio de __case object__

```scala
case object Conectado
case object Desconectado
case object EnMantenimiento

val estadoServidor = Conectado

println(estadoServidor)

def mensajeEstado(estado: Any): String = estado match {
  case Conectado       => "El servidor está disponible."
  case Desconectado    => "El servidor está apagado."
  case EnMantenimiento => "El servidor está en mantenimiento."
  case _               => "Estado desconocido."
}

println(mensajeEstado(estadoServidor))
println(mensajeEstado(EnMantenimiento))
```

### Explicación Línea a Línea

- **`case object Conectado`**  
  Declara un `case object` singleton llamado `Conectado`. No tiene datos, solo representa un estado concreto.

- **`case object Desconectado`**  
  Otro estado posible del servidor, también como singleton.

- **`case object EnMantenimiento`**  
  Tercer estado posible, sin datos asociados.

- **`val estadoServidor = Conectado`**  
  Asigna el estado actual del servidor al valor `Conectado`.

- **`println(estadoServidor)`**  
  Imprime `Conectado` gracias al `toString` automático de `case object`.

- **`def mensajeEstado(estado: Any): String = estado match {`**  
  Define una función que recibe un valor de tipo `Any` y devuelve un `String`. Usa pattern matching para decidir el mensaje.

- **`case Conectado       => "El servidor está disponible."`**  
  Si el valor es exactamente el singleton `Conectado`, devuelve este mensaje.

- **`case Desconectado    => "El servidor está apagado."`**  
  Mensaje para el estado `Desconectado`.

- **`case EnMantenimiento => "El servidor está en mantenimiento."`**  
  Mensaje específico para `EnMantenimiento`.

- **`case _               => "Estado desconocido."`**  
  Caso comodín para cualquier otro valor que no coincida con los anteriores.

- **`println(mensajeEstado(estadoServidor))`**  
  Llama a la función con `Conectado`, imprime el mensaje correspondiente.

- **`println(mensajeEstado(EnMantenimiento))`**  
  Llama a la función con `EnMantenimiento`, imprime el mensaje de mantenimiento.
